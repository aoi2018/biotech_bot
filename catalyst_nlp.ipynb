```python
import numpy as np
import pandas as pd
import gensim.downloader as api
import re
```

```python
nlp_model = api.load('glove-wiki-gigaword-300')
```

```python
df = pd.read_csv('datasets/catalyst_ds.csv')
```

```python
stop_words =('to', 'as', 'on', 'by', 'new', 'is', 'the', 'and', 'in', 'a', 'at', 'for', 'in', 'from', 'with', 'after')

def preprocessing(text):
    cleaned = re.sub(r'[^\w\s]', "", str(text).lower())
    tokens = cleaned.split()
    return[word for word in tokens if word not in stop_words and len(word)>1]
```

```python
def get_vector(tokens):
    vectors = [nlp_model[w] for w in tokens if w in nlp_model]
    return np.mean(vectors, axis = 0)  if vectors else np.zeros(300)
```

```python
cleaned_tokens = df['catalyst'].apply(preprocessing)
```

```python
polarity = np.vstack(cleaned_tokens.apply(get_vector).values)
```

```python
df['polarity'] = polarity.mean(axis=1)
```

```python
df.to_csv("datasets/catalyst_ds.csv", index = False)
```
