```python
import pandas as pd
from transformers import pipeline
import torch
```

```python
df = pd.read_csv("datasets/catalyst_ds.csv")
```

```python
finbert = pipeline(task = "text-classification",
                   model = "ProsusAI/finbert",
                   tokenizer = "ProsusAI/finbert",
                   )
```

```python
output = finbert(df['catalyst'].astype(str).to_list(), batch_size = 32, truncation = True)
```

```python
df['polarity'] = [opt['label'] for opt in output]
```

```python
df.to_csv("datasets/catalyst_ds.csv", index = False)
```
