### run sentiment polarity extractor

```python
import pandas as pd
from transformers import pipeline
import torch
```

```python
df = pd.read_csv("datasets/catalyst_ds.csv")
```

```python
finbert = pipeline(task = "text-classification",
                   model = "ProsusAI/finbert",
                   tokenizer = "ProsusAI/finbert",
                   )
```

```python
output = finbert(df['catalyst'].astype(str).to_list(), batch_size = 32, truncation = True, top_k = None)
```

```python
df['polarity'] = [{s['label']:s['score'] for s in opt}['positive']-
{s['label']:s['score'] for s in opt}['negative']
for opt in output]
```

```python
df.to_csv("datasets/catalyst_ds.csv", index = False)
```
