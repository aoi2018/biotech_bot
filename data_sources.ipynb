```python
# import libraries (pip install dependencies)
import pandas as pd
import yfinance as yf
```

```python
# define tickers of interest
tickers = ['VRTX', 'AMGN', 'BIIB']
```

```python
# pull financial data for each ticker
def getFinancialInfo(tickers):
    financial_data = []
    for ticker in tickers:
        company = yf.Ticker(ticker)
        info = company.info
        info['ticker'] = ticker
        financial_data.append(info)
```

```python
    df = pd.DataFrame(financial_data)

    keptColumns = ['ticker',
            'shortName', 
            'marketCap', 
            'totalRevenue', 
            'grossProfits', 
            'ebitda', 
            'totalCash', 
            'totalDebt', 
            'currentRatio', 
            'quickRatio', 
            'operatingCashflow']


    for col in keptColumns:
        if col not in df.columns:
            df[col] = None

    return df[keptColumns]
```
