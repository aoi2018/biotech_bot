```python
# import libraries (pip install dependencies)
import pandas as pd
import yfinance as yf
```

```python
# define tickers of interest
tickers = ['VRTX', 'AMGN', 'BIIB']
```

```python
# pull financial data for each ticker
def getFinancialInfo(tickers):
    financial_data = []
    for ticker in tickers:
        company = yf.Ticker(ticker)
        info = company.info
        info['ticker'] = ticker
        financial_data.append(info)
    return financial_data
```

```python
df = pd.DataFrame(getFinancialInfo(tickers))

keptColumns = ['ticker',
    'shortName', 
    'marketCap', 
    'totalRevenue', 
    'grossProfits', 
    'ebitda', 
    'totalCash', 
    'totalDebt', 
    'currentRatio', 
    'quickRatio', 
    'operatingCashflow']

company_features = df.reindex(columns = keptColumns)
company_features.columns
```

```python
market_data = []    
for ticker in tickers:
    data = yf.Ticker(ticker).history(start = '2021-01-01', end = '2026-01-01').reset_index()
    data['Ticker'] = ticker
    keptColumns = ['Ticker', 'Open', 'Close', 'High', 'Low', 'Volume']
    data = data.reindex(columns = keptColumns)
    market_data.append(data)
market_data = pd.concat(market_data, ignore_index=True)
market_data.shape
market_data.head()
```
